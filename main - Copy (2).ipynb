Many people don’t know they have heart disease until they suffer a heart attack or stroke.  

Most people are unwilling to visit doctors regularly to check on their health.  

**Our goal** is to **simplify** this process and enable people to check their health status without always needing to visit doctors.  

**The Problem**: We need a simple, accessible way to predict if someone is at risk for heart disease using basic health information.

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
```

```python
x = df1.drop("HeartDiseaseorAttack" , axis=1)
y = df1["HeartDiseaseorAttack"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print("Before SMOTE:\n", y_train.value_counts())
```

```python
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
print("After SMOTE:\n", y_train_res.value_counts())
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# بناء الموديل
model = xgb.XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

# تدريب الموديل على البيانات بعد SMOTE
print("Start Modeling ")
model.fit(X_train_res, y_train_res)
print("End Modeling ✅" )

# التنبؤ
y_pred = model.predict(X_test)

# النتائج
accuracy = accuracy_score(y_test, y_pred)
print(f"\n🎯 دقة الموديل: {accuracy:.2%}")
print("\n📊 تقرير التصنيف:")
print(classification_report(y_test, y_pred))

# رسم 1: Confusion Matrix
plt.figure(figsize=(8, 6))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.tight_layout()
plt.show()

# رسم 2: أهم الخصائص (Feature Importance)
plt.figure(figsize=(10, 6))
feature_importance = model.feature_importances_
features = x.columns
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': feature_importance
}).sort_values('Importance', ascending=False)

plt.barh(importance_df['Feature'][:10], importance_df['Importance'][:10], color='skyblue')
plt.xlabel('Importance', fontsize=12)
plt.title('أهم 10 خصائص في التنبؤ', fontsize=16, fontweight='bold')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# رسم 3: مقارنة بين Training و Testing
train_pred = model.predict(X_train_res)
train_acc = accuracy_score(y_train_res, train_pred)
test_acc = accuracy

plt.figure(figsize=(8, 6))
accuracies = [train_acc, test_acc]
labels = ['Training', 'Testing']
colors = ['lightcoral', 'lightgreen']

bars = plt.bar(labels, accuracies, color=colors, alpha=0.7, edgecolor='black')
plt.ylim(0, 1.1)
plt.ylabel('Accuracy', fontsize=12)
plt.title('دقة الموديل على Training و Testing', fontsize=16, fontweight='bold')

# إضافة الأرقام فوق الأعمدة
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02, 
             f'{acc:.2%}', ha='center', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n✨ خلصنا! الموديل جاهز للاستخدام")
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# بناء الموديل مع تحسينات
model = xgb.XGBClassifier(
    n_estimators=200,           # زودنا عدد الأشجار
    learning_rate=0.05,         # قللنا السرعة عشان يتعلم أحسن
    max_depth=6,                # زودنا العمق شوية
    min_child_weight=1,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=10,        # ⭐ ده أهم حاجة! بيركز على الحالات المريضة
    random_state=42
)

# تدريب الموديل
print("🚀 بدأ التدريب المحسّن...")
model.fit(X_train_res, y_train_res)
print("✅ التدريب خلص!")

# التنبؤ
y_pred = model.predict(X_test)

# النتائج
accuracy = accuracy_score(y_test, y_pred)
print(f"\n🎯 دقة الموديل: {accuracy:.2%}")
print("\n📊 تقرير التصنيف:")
print(classification_report(y_test, y_pred))

# رسم 1: Confusion Matrix محسّن
plt.figure(figsize=(10, 8))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='RdYlGn', cbar=True, 
            annot_kws={'size': 14, 'weight': 'bold'})
plt.title('Confusion Matrix (محسّن)', fontsize=18, fontweight='bold', pad=20)
plt.ylabel('الحقيقة (Actual)', fontsize=14)
plt.xlabel('التنبؤ (Predicted)', fontsize=14)

# إضافة نسب مئوية
for i in range(2):
    for j in range(2):
        text = plt.text(j + 0.5, i + 0.7, 
                       f'{cm[i,j]/cm[i].sum()*100:.1f}%',
                       ha="center", va="center", color="navy", fontsize=11)
plt.tight_layout()
plt.show()

# رسم 2: أهم الخصائص
plt.figure(figsize=(12, 8))
feature_importance = model.feature_importances_
features = x.columns
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': feature_importance
}).sort_values('Importance', ascending=False)

colors = plt.cm.viridis(importance_df['Importance'][:10] / importance_df['Importance'][:10].max())
plt.barh(importance_df['Feature'][:10], importance_df['Importance'][:10], color=colors)
plt.xlabel('أهمية الخاصية', fontsize=14, fontweight='bold')
plt.title('أهم 10 خصائص في اكتشاف مرض القلب ❤️', fontsize=16, fontweight='bold', pad=20)
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# رسم 3: مقارنة Precision & Recall للحالتين
from sklearn.metrics import precision_score, recall_score

precision_0 = precision_score(y_test, y_pred, pos_label=0)
recall_0 = recall_score(y_test, y_pred, pos_label=0)
precision_1 = precision_score(y_test, y_pred, pos_label=1)
recall_1 = recall_score(y_test, y_pred, pos_label=1)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# للأصحاء
categories = ['Precision', 'Recall']
values_healthy = [precision_0, recall_0]
ax[0].bar(categories, values_healthy, color=['lightblue', 'lightcoral'], alpha=0.8, edgecolor='black')
ax[0].set_ylim(0, 1.1)
ax[0].set_title('الأصحاء (No Heart Disease) 💚', fontsize=14, fontweight='bold')
ax[0].set_ylabel('Score', fontsize=12)
for i, v in enumerate(values_healthy):
    ax[0].text(i, v + 0.02, f'{v:.2%}', ha='center', fontsize=12, fontweight='bold')

# للمرضى
values_sick = [precision_1, recall_1]
ax[1].bar(categories, values_sick, color=['lightgreen', 'salmon'], alpha=0.8, edgecolor='black')
ax[1].set_ylim(0, 1.1)
ax[1].set_title('المرضى (Heart Disease) ❤️', fontsize=14, fontweight='bold')
ax[1].set_ylabel('Score', fontsize=12)
for i, v in enumerate(values_sick):
    ax[1].text(i, v + 0.02, f'{v:.2%}', ha='center', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# رسم 4: Training vs Testing
train_pred = model.predict(X_train_res)
train_acc = accuracy_score(y_train_res, train_pred)
test_acc = accuracy

plt.figure(figsize=(10, 6))
x_pos = [0, 1]
accuracies = [train_acc, test_acc]
labels = ['Training\nالتدريب', 'Testing\nالاختبار']
colors = ['#FF6B6B', '#4ECDC4']

bars = plt.bar(x_pos, accuracies, color=colors, alpha=0.8, edgecolor='black', linewidth=2)
plt.ylim(0, 1.1)
plt.ylabel('Accuracy (الدقة)', fontsize=14, fontweight='bold')
plt.title('أداء الموديل 📊', fontsize=18, fontweight='bold', pad=20)
plt.xticks(x_pos, labels, fontsize=12)
plt.grid(axis='y', alpha=0.3, linestyle='--')

for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.03, 
             f'{acc:.2%}', ha='center', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n" + "="*50)
print("✨ التحليل النهائي:")
print("="*50)
print(f"📈 Recall للمرضى: {recall_1:.2%} (قد إيه بنكتشف المرضى)")
print(f"🎯 Precision للمرضى: {precision_1:.2%} (قد إيه تنبؤاتنا صح)")
print("="*50)
```

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 🎯 الموديل المحسّن بأفضل Parameters
model = xgb.XGBClassifier(
    n_estimators=300,              # زودنا عدد الأشجار
    learning_rate=0.03,            # تعلم أبطأ = دقة أعلى
    max_depth=7,                   # عمق أكبر للتعلم الأفضل
    min_child_weight=1,
    subsample=0.85,                # استخدام 85% من البيانات في كل شجرة
    colsample_bytree=0.85,         # استخدام 85% من الخصائص
    gamma=0.1,                     # منع Overfitting
    reg_alpha=0.1,                 # L1 regularization
    reg_lambda=1,                  # L2 regularization
    scale_pos_weight=6,            # توازن ممتاز
    random_state=42,
    eval_metric='logloss'
)

print("🚀 بدأ التدريب المحسّن جداً...")
print("⏳ الموضوع ممكن ياخد دقيقة...")

# تدريب مع Early Stopping
eval_set = [(X_test, y_test)]
model.fit(
    X_train_res, y_train_res,
    eval_set=eval_set,
    verbose=False
)

print("✅ التدريب خلص!")

# التنبؤ
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)[:, 1]

# النتائج الأساسية
accuracy = accuracy_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("\n" + "="*60)
print("🎯 النتائج النهائية:")
print("="*60)
print(f"📊 دقة الموديل (Accuracy): {accuracy:.2%}")
print(f"🎪 ROC-AUC Score: {roc_auc:.3f}")
print("="*60)
print("\n📋 تقرير التصنيف التفصيلي:")
print(classification_report(y_test, y_pred, target_names=['أصحاء', 'مرضى']))

# Cross Validation للتأكد من استقرار الموديل
print("\n🔄 جاري Cross Validation...")
cv_scores = cross_val_score(model, X_train_res, y_train_res, cv=5, scoring='recall')
print(f"✅ متوسط Recall في CV: {cv_scores.mean():.2%} (±{cv_scores.std():.2%})")

# ==========================
# الرسومات المحسّنة
# ==========================

# رسم 1: Confusion Matrix فخم
fig = plt.figure(figsize=(12, 10))
gs = fig.add_gridspec(2, 2, hspace=0.3, wspace=0.3)

ax1 = fig.add_subplot(gs[0, 0])
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='RdYlGn', cbar=True, 
            annot_kws={'size': 16, 'weight': 'bold'}, ax=ax1,
            xticklabels=['أصحاء', 'مرضى'], yticklabels=['أصحاء', 'مرضى'])
ax1.set_title('Confusion Matrix 🎯', fontsize=16, fontweight='bold', pad=15)
ax1.set_ylabel('الحقيقة', fontsize=13, fontweight='bold')
ax1.set_xlabel('التنبؤ', fontsize=13, fontweight='bold')

# إضافة النسب
for i in range(2):
    for j in range(2):
        ax1.text(j + 0.5, i + 0.7, f'{cm[i,j]/cm[i].sum()*100:.1f}%',
                ha="center", va="center", color="navy", fontsize=11, weight='bold')

# رسم 2: ROC Curve
ax2 = fig.add_subplot(gs[0, 1])
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
ax2.plot(fpr, tpr, linewidth=3, label=f'ROC Curve (AUC = {roc_auc:.3f})', color='#e74c3c')
ax2.plot([0, 1], [0, 1], 'k--', linewidth=2, label='Random Classifier')
ax2.fill_between(fpr, tpr, alpha=0.3, color='#e74c3c')
ax2.set_xlabel('False Positive Rate', fontsize=12, fontweight='bold')
ax2.set_ylabel('True Positive Rate', fontsize=12, fontweight='bold')
ax2.set_title('ROC Curve 📈', fontsize=16, fontweight='bold', pad=15)
ax2.legend(loc='lower right', fontsize=10)
ax2.grid(alpha=0.3)

# رسم 3: Feature Importance
ax3 = fig.add_subplot(gs[1, :])
feature_importance = model.feature_importances_
features = x.columns
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': feature_importance
}).sort_values('Importance', ascending=False).head(12)

colors = plt.cm.plasma(np.linspace(0.3, 0.9, len(importance_df)))
bars = ax3.barh(importance_df['Feature'], importance_df['Importance'], color=colors, edgecolor='black', linewidth=1.5)
ax3.set_xlabel('أهمية الخاصية', fontsize=13, fontweight='bold')
ax3.set_title('أهم 12 خاصية في التنبؤ بمرض القلب ❤️', fontsize=16, fontweight='bold', pad=15)
ax3.invert_yaxis()
ax3.grid(axis='x', alpha=0.3, linestyle='--')

# إضافة القيم على الأعمدة
for bar in bars:
    width = bar.get_width()
    ax3.text(width, bar.get_y() + bar.get_height()/2, f'{width:.3f}',
            ha='left', va='center', fontsize=9, weight='bold')

plt.tight_layout()
plt.show()

# ==========================
# رسم 4: مقارنة شاملة
# ==========================
from sklearn.metrics import precision_score, recall_score, f1_score

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# للأصحاء
metrics_healthy = {
    'Precision': precision_score(y_test, y_pred, pos_label=0),
    'Recall': recall_score(y_test, y_pred, pos_label=0),
    'F1-Score': f1_score(y_test, y_pred, pos_label=0)
}
bars1 = axes[0].bar(metrics_healthy.keys(), metrics_healthy.values(), 
                    color=['#3498db', '#2ecc71', '#f39c12'], alpha=0.8, edgecolor='black', linewidth=2)
axes[0].set_ylim(0, 1.1)
axes[0].set_title('أداء التنبؤ للأصحاء 💚', fontsize=14, fontweight='bold', pad=15)
axes[0].set_ylabel('Score', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)
for bar in bars1:
    height = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{height:.1%}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# للمرضى
metrics_sick = {
    'Precision': precision_score(y_test, y_pred, pos_label=1),
    'Recall': recall_score(y_test, y_pred, pos_label=1),
    'F1-Score': f1_score(y_test, y_pred, pos_label=1)
}
bars2 = axes[1].bar(metrics_sick.keys(), metrics_sick.values(), 
                    color=['#e74c3c', '#e67e22', '#9b59b6'], alpha=0.8, edgecolor='black', linewidth=2)
axes[1].set_ylim(0, 1.1)
axes[1].set_title('أداء التنبؤ للمرضى ❤️', fontsize=14, fontweight='bold', pad=15)
axes[1].set_ylabel('Score', fontsize=12, fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)
for bar in bars2:
    height = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{height:.1%}', ha='center', va='bottom', fontsize=11, fontweight='bold')

# مقارنة Train vs Test
train_pred = model.predict(X_train_res)
train_acc = accuracy_score(y_train_res, train_pred)
train_recall = recall_score(y_train_res, train_pred)
test_acc = accuracy
test_recall = recall_score(y_test, y_pred)

x = np.arange(2)
width = 0.35
bars3 = axes[2].bar(x - width/2, [train_acc, train_recall], width, 
                    label='Training', color='#16a085', alpha=0.8, edgecolor='black', linewidth=2)
bars4 = axes[2].bar(x + width/2, [test_acc, test_recall], width,
                    label='Testing', color='#c0392b', alpha=0.8, edgecolor='black', linewidth=2)
axes[2].set_ylabel('Score', fontsize=12, fontweight='bold')
axes[2].set_title('Training vs Testing 📊', fontsize=14, fontweight='bold', pad=15)
axes[2].set_xticks(x)
axes[2].set_xticklabels(['Accuracy', 'Recall'], fontsize=11)
axes[2].legend(fontsize=10)
axes[2].set_ylim(0, 1.1)
axes[2].grid(axis='y', alpha=0.3)

for bars in [bars3, bars4]:
    for bar in bars:
        height = bar.get_height()
        axes[2].text(bar.get_x() + bar.get_width()/2., height + 0.02,
                    f'{height:.1%}', ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

# ==========================
# الملخص النهائي
# ==========================
print("\n" + "🌟"*30)
print("✨ ملخص الأداء النهائي:")
print("🌟"*30)
print(f"🎯 Accuracy: {accuracy:.2%}")
print(f"📈 Recall للمرضى: {recall_score(y_test, y_pred):.2%}")
print(f"🎪 Precision للمرضى: {precision_score(y_test, y_pred):.2%}")
print(f"⚡ F1-Score للمرضى: {f1_score(y_test, y_pred):.2%}")
print(f"🏆 ROC-AUC: {roc_auc:.3f}")
print("🌟"*30)
print("\n💪 الموديل جاهز للاستخدام!")
```
