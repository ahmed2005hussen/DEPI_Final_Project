## 📝 Project Statement

Heart disease is one of the most **common** diseases in the world which often leads to death.  

Many people don’t know they have heart disease until they suffer a heart attack or stroke.  

Most people are unwilling to visit doctors regularly to check on their health.  

**Our goal** is to **simplify** this process and enable people to check their health status without always needing to visit doctors.  

**The Problem**: We need a simple, accessible way to predict if someone is at risk for heart disease using basic health information.

---

## BMI and Heart  disease

```python
bins = [0, 15, 20, 25, 30, 35, 100]
labels = ['<15', '15-20', '20-25', '25-30', '30-35', '35+']
df['BMI_range'] = pd.cut(df['BMI'], bins=bins, labels=labels, right=False)

plt.figure(figsize=(10,6))
sns.countplot(
    data=df,
    x='BMI_range',
    hue='HeartDiseaseorAttack',
    palette='Set1'
)

plt.title("BMI Ranges vs Heart Disease/Attack", fontsize=14)
plt.xlabel("BMI Range")
plt.ylabel("Count")
plt.legend(title="Heart Disease/Attack", labels=["No", "Yes"])
plt.show()
```

Most individuals fall within the BMI ranges of 20–30. However, the proportion of people with Heart Disease increases noticeably in higher BMI ranges (30+).

This suggests a positive relationship between obesity and the likelihood of heart disease or attack.

# BMI Box plot

```python
sns.boxplot(data = df, x='BMI')

median = df["BMI"].quantile(0.50)
q1 = df["BMI"].quantile(0.25)
q3 = df["BMI"].quantile(0.75)
IQR = q3 - q1

lowerWisker = q1 -  1.5 * IQR
upperWisker = q3 + 1.5 * IQR


print(f"quantile 1 = {q1}")
print(f"quantile 3 = {q3}")
print(f"median = {median}")
print(f"IQR = {IQR}")
print(f"lowerwisker = {lowerWisker}")
print(f"upperWisker = {upperWisker}")
print("\n")

plt.title("Box plot for BMI")

plt.show()
```

The box plot for BMI shows that while most individuals fall between Q1 ≈ 24 and Q3 ≈ 32, there are many outliers above the upper whisker (~40). These outliers represent individuals with significantly higher BMI values, and they are important to retain since they capture cases of severe obesity, which are strongly linked to health risks such as heart disease, diabetes, and high blood pressure.

# Smoking and Heart Disease

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Smoker", hue="HeartDiseaseorAttack")
plt.title("Heart Disease vs Smoker")
plt.xlabel("Smoker")
plt.ylabel("Count")
plt.legend(title="Heart Disease", labels=["No", "Yes"])
plt.show()
```

### 1. Comparing smokers vs non-smokers

Among non-smokers (0):

The majority are free of heart disease.

Only a relatively small portion have heart disease.

Among smokers (1):

The total number of people is lower than non-smokers, but the share of heart disease cases is visibly higher compared to non-smokers.

This suggests that smoking is positively associated with heart disease.

<br>

### 2. Insight for health risk

Even though most smokers do not have heart disease, the risk increases compared to non-smokers.

This matches medical research: smoking damages blood vessels and increases the risk of cardiovascular disease.

## Conclusion:
Smoking appears to be a risk factor for heart disease in your dataset. While the absolute number of heart disease cases is smaller than non-cases, smokers have a higher relative proportion of heart disease than non-smokers.

# Physical Activity and Heart Disease

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="PhysActivity", hue="HeartDiseaseorAttack")
plt.title("Heart Disease vs Physical Activity")
plt.xlabel("Physical Activity")
plt.ylabel("Count")
plt.legend(title="Heart Disease", labels=["No", "Yes"])
plt.show()
```

### 1. General distribution

Both groups (those who do physical activity = 1, and those who do not = 0) have a much higher count of people without heart disease compared to those with it.

This aligns with the fact that heart disease is relatively less common in the general population.

<br>

### 2. Comparing physical activity vs no physical activity

Physically active group (1):

The number of people without heart disease is very large.

The proportion of people with heart disease is smaller compared to the inactive group.

Physically inactive group (0):

The proportion of people with heart disease is noticeably higher compared to the active group.

Even though the inactive group may have fewer people overall, their relative share of heart disease is higher.

<br>

### 3. Health insight

Physical activity seems to have a protective effect against heart disease in your dataset.

Lack of physical activity is correlated with a higher risk of heart disease.

This reinforces the widely known medical fact that regular physical activity strengthens the cardiovascular system and reduces heart disease risk.

## Conclusion:
People who engage in physical activity (PhysActivity = 1) are less likely to have heart disease compared to those who don’t (PhysActivity = 0). The plot strongly suggests that maintaining an active lifestyle is a key protective factor.
